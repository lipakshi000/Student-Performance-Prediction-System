```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#import dataset
df=pd.read_csv("student_info.csv")
df.head()
```

```python
df.shape
```

```python
#check missing values
df.isnull().sum()
```

```python
#handle missing values
df=df.fillna(df.mean())
df.isnull().sum()
```

```python
#Visualization
plt.scatter(x=df.study_hours,y=df.student_marks)
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Student Marks Predictor")
plt.show()
```

```python
#seperate dependent and independent varibales
X=df.drop("student_marks",axis=1)
y=df.drop("study_hours",axis=1)  # (axis=1 clearifies droping the column for columns axis=1 , for rows axis =0 ) 
```

```python
X.head()
```

```python
y.head()
```

```python
#train_test_spilitting
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
# apply linear regression algo....
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(X_train,y_train)
```

```python
#test the model
res=lr.predict([[8.0]])[0][0].round(2)
print(res)
```

```python
#how to create the .PKL file , as extract of this model
import pickle
pickle.dump(lr,open("smp.pkl","wb"))       #"wc=eerite binary"   ,"smp=student mark predictor"
print("Model is saved")
```
